```python
# GPU check — need Compute Capability >= 8.0 (A100 / L4). T4 (7.5) cannot run Ozaki MMA.
import subprocess, sys
out = subprocess.check_output(
    ["nvidia-smi", "--query-gpu=name,compute_cap", "--format=csv,noheader"],
    text=True,
).strip().splitlines()[0]
print(out)
cap = float(out.split(",")[-1].strip())
if cap < 8.0:
    raise RuntimeError(
        f"GPU {out} is CC {cap} (e.g. T4). Need CC >= 8.0 (A100/L4). "
        "Colab: Runtime -> Change runtime type -> GPU -> A100"
    )
print("OK: GPU supports INT8 m16n8k32 Ozaki kernels.")
```
